# 4. Step 1 - Data Cleaning

Data cleaning is the foundation of any reliable modelling pipeline. Poor cleaning leads to inflated CV scores, biased coefficients, and models that fail in production. The steps in this section follow standard practice:

1. **Missing values** - detect and, if present, handle via imputation or removal.
2. **Duplicates** - detect and drop, because duplicated rows inflate the effective training signal and can leak between train and test splits.
3. **Identifier columns** - drop non-predictive columns (such as `employee_id`) so they cannot mislead the encoder or the model.
4. **Outliers** - detect using the IQR rule plus a skewness test, then treat in one of three ways based on distribution shape:
   - Highly skewed features: IQR capping to preserve the tail's information while bounding its influence.
   - Roughly symmetric features: Winsorisation at the IQR fences.
   - A second-round 5th/95th percentile cap for any column that still shows fence violations after the first pass.

This tiered approach is preferred over blanket removal because it retains the full 50,000-row sample size while controlling the influence of extreme observations on scale-sensitive learners.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split, cross_val_score, KFold, GridSearchCV
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.neighbors import KNeighborsRegressor
from sklearn.ensemble import VotingRegressor
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from xgboost import XGBRegressor
import warnings
warnings.filterwarnings("ignore")

df = pd.read_csv("remote_work_productivity_dataset.csv")
print("="*60)
print("DATASET OVERVIEW")
print("="*60)
print(f"Rows: {df.shape[0]}   Columns: {df.shape[1]}")

print("\n" + "="*60)
print("MISSING VALUES ANALYSIS")
print("="*60)
missing = df.isnull().sum().sum()
if missing > 0:
    print(f"Missing Values Found: {missing}")
    for col in df.columns:
        mv = df[col].isnull().sum()
        if mv > 0:
            print(f"{col}: {mv}")
else:
    print("No Missing Values Found")

print("\n" + "="*60)
print("DUPLICATE RECORDS ANALYSIS")
print("="*60)
duplicates = df.duplicated().sum()
if duplicates > 0:
    print(f"Duplicate Rows Found: {duplicates}")
    df = df.drop_duplicates().reset_index(drop=True)
    print("Duplicates removed.")
else:
    print("No Duplicate Rows Found")

drop_cols = ['employee_id', 'Employee_ID', 'ID', 'Name']
for col in drop_cols:
    if col in df.columns:
        df.drop(col, axis=1, inplace=True)
        print(f"Dropped identifier column: {col}")

print("\n" + "="*60)
print("OUTLIER DETECTION REPORT")
print("="*60)
numeric_cols = df.select_dtypes(include=np.number).columns
outlier_summary = []
for col in numeric_cols:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    outliers = ((df[col] < lower) | (df[col] > upper)).sum()
    if outliers > 0:
        skewness = df[col].skew()
        if abs(skewness) > 1:
            outlier_type = "Skewed Distribution Outliers"
            action = "Recommend Capping / Transformation"
        else:
            outlier_type = "Mild / Normal Distribution Outliers"
            action = "Recommend Winsorization / Removal"
        print(f"{col}")
        print(f"   Outliers Found : {outliers}")
        print(f"   Skewness       : {round(skewness,3)}")
        print(f"   Type           : {outlier_type}")
        print(f"   Suggested Fix  : {action}\n")
        outlier_summary.append([col, outliers, round(skewness,3), outlier_type])
    else:
        print(f"{col:<25} No Outliers")

print("\n" + "="*60)
print("FINAL SUMMARY")
print("="*60)
if len(outlier_summary) > 0:
    print("Columns Containing Outliers:\n")
    for row in outlier_summary:
        print(f"{row[0]} | Count: {row[1]} | Skewness: {row[2]} | {row[3]}")
else:
    print("No Outliers Detected In Dataset")
```

## 4.1 Outlier Treatment - First Pass

Outliers are treated in place on a copy of the dataframe (`cleaned_df`) rather than dropped, to preserve sample size. Two strategies are used, chosen automatically based on each column's skewness:

- **|skewness| > 1** - IQR capping via `np.clip`. This pulls extreme tail values to the 1.5 x IQR fence, which is the standard non-parametric bound for skewed data.
- **|skewness| <= 1** - Winsorisation at the same IQR fences. Mathematically similar, but framed as symmetric trimming for roughly normal columns.

Integer-valued score columns (`tasks_completed`, `mental_health_score`, `job_satisfaction`) are rounded after capping to preserve their semantic integer type.


```python
print("OUTLIER TREATMENT STARTED")
cleaned_df = df.copy()
for col in numeric_cols:
    q1 = cleaned_df[col].quantile(0.25)
    q3 = cleaned_df[col].quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    outliers = ((cleaned_df[col] < lower) | (cleaned_df[col] > upper)).sum()
    if outliers > 0:
        skewness = cleaned_df[col].skew()
        print(f"\nColumn: {col}")
        print(f"Outliers Before Treatment: {outliers}")
        if abs(skewness) > 1:
            cleaned_df[col] = np.clip(cleaned_df[col], lower, upper)
            print(f"Treatment Applied: IQR Capping")
            print(f"Reason: Highly Skewed Distribution")
        else:
            cleaned_df[col] = np.where(
                cleaned_df[col] < lower,
                lower,
                cleaned_df[col]
            )
            cleaned_df[col] = np.where(
                cleaned_df[col] > upper,
                upper,
                cleaned_df[col]
            )
            print("Treatment Applied: Winsorization")
            print("Reason: Mild / Normal Distribution")
        if col in [
            "tasks_completed",
            "mental_health_score",
            "job_satisfaction"
        ]:
            cleaned_df[col] = round(cleaned_df[col])
print(f"OUTLIER RECHECK REPORT")
for col in numeric_cols:
    q1 = cleaned_df[col].quantile(0.25)
    q3 = cleaned_df[col].quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    remaining = ((cleaned_df[col] < lower) | (cleaned_df[col] > upper)).sum()
    if remaining == 0:
        print(f"{col:<30} Cleaned Successfully")
    else:
        print(f"{col:<30} Remaining Outliers: {remaining}")
print(f"Cleaned dataframe stored in variable: cleaned_df")
```

## 4.2 Outlier Treatment - Second Pass (`job_satisfaction`)

After the first pass, `job_satisfaction` may still show IQR-fence violations because the initial clip recalculated its quartiles on the already-clipped data, leaving a residual boundary effect. A stronger 5th-95th percentile cap is applied to force convergence. This percentile-based method is more aggressive than IQR capping and is appropriate here because `job_satisfaction` is a bounded ordinal score where extreme tails carry little additional predictive signal.


```python
print(f"SECOND ROUND TREATMENT : job_satisfaction")
col = "job_satisfaction"
lower = cleaned_df[col].quantile(0.05)
upper = cleaned_df[col].quantile(0.95)
before = (
    (cleaned_df[col] < lower) |
    (cleaned_df[col] > upper)
).sum()
print(f"Outliers Before Second Treatment: {before}")
cleaned_df[col] = np.clip(cleaned_df[col], lower, upper)
# Round values
cleaned_df[col] = round(cleaned_df[col])
print(f"Treatment Applied: 5th-95th Percentile Capping")
q1 = cleaned_df[col].quantile(0.25)
q3 = cleaned_df[col].quantile(0.75)
iqr = q3 - q1
lower_iqr = q1 - 1.5 * iqr
upper_iqr = q3 + 1.5 * iqr

remaining = (
    (cleaned_df[col] < lower_iqr) |
    (cleaned_df[col] > upper_iqr)
).sum()
print(f"Remaining Outliers After Treatment: {remaining}")
if remaining == 0:
    print("job_satisfaction Cleaned Successfully")
```

## 4.3 Final Outlier Recheck

A final full-dataset recheck confirms the pipeline has converged. Any column reporting zero remaining outliers has satisfied the IQR criterion, which is what we expect after the two-pass treatment above.


```python
print("="*60)
print("OUTLIER RECHECK REPORT")
print("="*60)
for col in numeric_cols:
    q1 = cleaned_df[col].quantile(0.25)
    q3 = cleaned_df[col].quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    remaining = ((cleaned_df[col] < lower) | (cleaned_df[col] > upper)).sum()
    if remaining == 0:
        print(f"{col:<30} Cleaned Successfully")
    else:
        print(f"{col:<30} Remaining Outliers: {remaining}")
print(f"\nCleaned dataframe stored in variable: cleaned_df")
```

# 5. Step 2 - Data Preprocessing

Two preprocessing steps are applied here:

- **Encoding:** Categorical columns (`country`, `industry`, `work_mode`, `burnout_level`) are converted to integer codes via `LabelEncoder`. Label encoding is chosen over one-hot encoding for two reasons: (i) tree-based models (Decision Tree, Random Forest, Gradient Boosting, XGBoost) handle label-encoded categoricals without the artificial ordinality problem that trips up linear models, and (ii) one-hot encoding `country` would expand the feature space by dozens of columns without a proportional information gain.
- **Scaling:** `StandardScaler` is applied to features before training the scale-sensitive learners (Linear Regression, KNN). Tree-based methods are invariant to feature scale, but linear and distance-based methods benefit from standardisation because it stabilises coefficient estimation (linear) and distance computation (KNN).

The data is split 80/20 for train/test using `random_state=42` for reproducibility. The scaler is fit on the training set only to prevent information leakage from test to train.


```python
print("="*60)
print("PREPROCESSING STARTED")
print("="*60)
df = cleaned_df.copy()
cat_cols = df.select_dtypes(include='object').columns
if len(cat_cols) > 0:
    le = LabelEncoder()
    for col in cat_cols:
        df[col] = le.fit_transform(df[col].astype(str))
        print(f"{col:<30} Encoded")
    print("\nCategorical Columns Encoded Successfully")
else:
    print("No Categorical Columns Found")

print("\n" + "="*60)
print("FEATURES AND TARGET")
print("="*60)
target = "productivity_score"
X = df.drop(target, axis=1)
y = df[target]
print(f"Features Shape : {X.shape}")
print(f"Target Shape   : {y.shape}")

print("\n" + "="*60)
print("TRAIN TEST SPLIT (80/20)")
print("="*60)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)
print(f"X_train Shape : {X_train.shape}")
print(f"X_test Shape  : {X_test.shape}")
print(f"y_train Shape : {y_train.shape}")
print(f"y_test Shape  : {y_test.shape}")

print("\n" + "="*60)
print("FEATURE SCALING")
print("="*60)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
print("Scaling Applied Successfully (fit on train only - no leakage)")
print(f"Scaled Train Shape : {X_train_scaled.shape}")
print(f"Scaled Test Shape  : {X_test_scaled.shape}")
```

# 6. Step 3 - Baseline Regression Models

Used three baseline (non-ensemble) regressors trained with appropriate hyperparameters and evaluated with RMSE, R-squared, and k-fold cross-validation.

## 6.1 Choice of Baselines

| Model | Why chosen as a baseline |
|---|---|
| **Linear Regression** | The simplest possible regression model. Gives an unregularised reference point and tells us how much of the target's variance is explained by a purely linear combination of features. Highly interpretable - coefficients map directly onto feature effects. |
| **K-Nearest Neighbors (KNN)** | A non-parametric distance-based baseline. Makes no assumption about the functional form and captures local structure that linear methods cannot. Provides a second angle on the data and a sanity check that does not share the linear-model assumptions. |
| **Decision Tree Regressor** | A non-linear, non-parametric baseline. Captures interactions and non-linearities that linear models miss. Acts as the natural comparison point for the tree-based ensembles (Gradient Boosting, XGBoost, Voting all contain trees). |

All three baselines are evaluated using **5-fold cross-validation** (k=5 is the standard bias/variance trade-off for k-fold CV) on R-squared, plus hold-out RMSE, R-squared, and MAE on the 20% test split.

## 6.2 Hyperparameter Choices for Baselines

| Model | Hyperparameter | Value | Rationale |
|---|---|---|---|
| Linear Regression | - | default | No hyperparameters to tune. Serves as the unregularised reference. |
| KNN | `n_neighbors` | 5 | Standard default; small enough to capture local structure on 40k training rows without excessive smoothing. |
| Decision Tree | `max_depth` | 10 | Limits overfitting on a 50k-row dataset while still allowing enough splits to capture interaction effects. |
| Decision Tree | `random_state` | 42 | Reproducibility. |


```python
baseline_grp = {
    "Linear Regression": LinearRegression(),
    "KNN": KNeighborsRegressor(n_neighbors=5),
    "Decision Tree": DecisionTreeRegressor(max_depth=10, random_state=42)
}

ensemble_grp = {
    "Gradient Boosting": GradientBoostingRegressor(
        n_estimators=200, learning_rate=0.05, random_state=42
    ),
    "XGBoost": XGBRegressor(
        n_estimators=250, learning_rate=0.05, max_depth=6,
        random_state=42, objective='reg:squarederror'
    ),
    "Voting Ensemble": VotingRegressor(estimators=[
        ('lr', LinearRegression()),
        ('dt', DecisionTreeRegressor(max_depth=10, random_state=42)),
        ('rf', RandomForestRegressor(n_estimators=150, max_depth=10, random_state=42))
    ])
}

kfold = KFold(n_splits=5, shuffle=True, random_state=42)

def process_models(model_dict, title):
    results = []
    for name, model in model_dict.items():
        if name in ["Linear Regression", "KNN"]:
            xtr, xte = X_train_scaled, X_test_scaled
        else:
            xtr, xte = X_train, X_test

        model.fit(xtr, y_train)
        pred = model.predict(xte)

        cv = cross_val_score(model, xtr, y_train, cv=kfold, scoring='r2', n_jobs=-1)

        results.append({
            "Model":    name,
            "R2 Score": round(r2_score(y_test, pred), 4),
            "RMSE":     round(np.sqrt(mean_squared_error(y_test, pred)), 4),
            "MAE":      round(mean_absolute_error(y_test, pred), 4),
            "CV Mean (R2)": round(cv.mean(), 4),
            "CV Std (R2)":  round(cv.std(), 4),
        })

    df = pd.DataFrame(results).sort_values(by="R2 Score", ascending=False).reset_index(drop=True)

    metrics_df = df.set_index("Model")[["R2 Score", "RMSE", "MAE", "CV Mean (R2)"]].T
    plt.figure(figsize=(12, 5))
    for col in metrics_df.columns:
        plt.plot(metrics_df.index, metrics_df[col], marker='o', label=col, linewidth=2)
    plt.title(f"{title} - Metric Comparison", fontsize=14, fontweight='bold')
    plt.grid(True, linestyle='--', alpha=0.6)
    plt.legend(loc='best')
    plt.tight_layout()
    plt.show()

    print(f"\n{title.upper()} REPORT")
    print(df.to_string(index=False))
    print(f"\nBest Model in this group: {df.iloc[0]['Model']}")
    return df

print("="*60)
print("ANALYZING BASELINE MODELS (non-ensemble)")
print("="*60)
df_base = process_models(baseline_grp, "Baseline Models (LR, KNN, DT)")
```

### 6.3 Baseline Comparison - Strengths and Weaknesses

A brief comparison of the three baselines:

- **Linear Regression** is the most interpretable model - each coefficient is a direct marginal effect on the scaled feature. It can only capture linear structure, so if `productivity_score` has meaningful non-linear dependence on `weekly_hours_worked`, `burnout_level`, or their interactions, OLS will underfit. Its value is primarily as an interpretability reference.
- **KNN** makes no functional-form assumption and can capture local patterns, but it scales poorly on large datasets (prediction time grows with training size) and suffers from the curse of dimensionality. Its cross-validation standard deviation is a useful indicator of how noisy the local neighbourhoods are.
- **Decision Tree** captures non-linearities and interactions that the other two baselines miss. However, a single tree is high-variance: small changes in the training data can produce very different trees. Its CV standard deviation is typically the largest of the three, which is exactly the motivation for moving to ensembles in Section 7.

Whichever baseline wins here defines the **bar the ensembles must clear** to justify their added complexity and loss of interpretability.


# 7. Step 3 - Ensemble Regression Models

## 7.1 Why Ensembles?

A single decision tree is high-variance: two trees trained on slightly different subsets of the same data can produce very different predictions. Ensembles address this by combining many base learners:

- **Bagging** (e.g. Random Forest) trains many trees on bootstrap resamples and averages their predictions. This drives variance down without materially increasing bias, so the ensemble generalises much better than any single tree.
- **Boosting** (e.g. Gradient Boosting, XGBoost) fits trees sequentially, where each new tree focuses on the residuals of the previous ensemble. This reduces bias in addition to variance and typically delivers the strongest predictive performance on tabular data.
- **Voting** combines structurally different learners so their errors decorrelate; the averaged prediction is more robust than any individual component.

For a regression problem on structured tabular data with mixed feature types - which is exactly what we have - tree ensembles are the canonical strong baseline and routinely win Kaggle competitions on similar data.

## 7.2 Ensemble Methods Selected and Justification

| Model | Family | Why included |
|---|---|---|
| **Gradient Boosting** | Boosting (sklearn) | Sequential residual-fitting; typically reduces bias compared to bagging. Serves as the standard boosting baseline. |
| **XGBoost** | Boosting (regularised) | Industry-standard boosted-tree implementation with built-in L1/L2 regularisation, sparsity handling, and strong out-of-the-box performance. Expected to be the strongest single model. |
| **Voting Regressor** | Heterogeneous averaging | Averages Linear Regression, Decision Tree, and Random Forest. Included to test whether combining *different* model families reduces error beyond any single family alone. |

## 7.3 Ensemble Hyperparameters (Initial)

| Model | Key hyperparameters | Initial values | Purpose |
|---|---|---|---|
| Gradient Boosting | `n_estimators`, `learning_rate` | 200, 0.05 | Trade-off between number of boosting rounds and per-round step size. |
| XGBoost | `n_estimators`, `learning_rate`, `max_depth` | 250, 0.05, 6 | Tree count, step size, and tree depth. `max_depth=6` is XGBoost's typical sweet spot for tabular data. |
| Voting Regressor | Base learners | LR + DT + RF | Uses default equal-weight averaging. |

These initial values are informed common-sense defaults. In Section 7.4 the strongest ensemble is refined further using GridSearchCV.


```python
print("="*60)
print("ANALYZING ENSEMBLE MODELS")
print("="*60)
df_ens = process_models(ensemble_grp, "Ensemble Models (GB, XGBoost, Voting)")
```

## 7.4 Hyperparameter Tuning with GridSearchCV

The strongest ensemble from Section 7.3 is refined further using `GridSearchCV`.

### Optimisation Strategy

- **Search algorithm:** Exhaustive grid search (`GridSearchCV`) over a deliberately small but impactful grid. Exhaustive search is preferred over random search here because the grid contains only 27 combinations or fewer, which is tractable in a few minutes of runtime.
- **Grid design:** The grid centres on the known sweet spot for tabular boosting - `max_depth` in 4-8, `learning_rate` in 0.03-0.1, `n_estimators` in 200-500. Literature on tabular boosting (Chen and Guestrin, 2016) shows this region captures the bulk of the tuning gain while avoiding combinatorial explosion.
- **Scoring metric:** R-squared, consistent with the baseline evaluation.
- **Cross-validation inside the grid:** 3-fold CV inside the search (each candidate is evaluated 3 times). This is a deliberate trade-off: 3 folds are coarser but keep total training from exploding (27 candidates x 3 folds = 81 model fits). The selected model is then re-validated with full 5-fold CV on the hold-out comparison below.
- **Parallelism:** `n_jobs=-1` uses all available CPU cores.


```python
best_ensemble_name = df_ens.iloc[0]["Model"]
print(f"Strongest ensemble from initial pass: {best_ensemble_name}")
print("Applying GridSearchCV to tune it further.\n")

if best_ensemble_name == "XGBoost":
    tuning_model = XGBRegressor(random_state=42, objective='reg:squarederror', n_jobs=-1)
    param_grid = {
        "n_estimators":  [200, 300, 500],
        "max_depth":     [4, 6, 8],
        "learning_rate": [0.03, 0.05, 0.1],
    }
elif best_ensemble_name == "Gradient Boosting":
    tuning_model = GradientBoostingRegressor(random_state=42)
    param_grid = {
        "n_estimators":  [150, 250, 400],
        "max_depth":     [3, 5, 7],
        "learning_rate": [0.03, 0.05, 0.1],
    }
else:
    tuning_model = RandomForestRegressor(random_state=42, n_jobs=-1)
    param_grid = {
        "n_estimators": [150, 300, 500],
        "max_depth":    [8, 12, 16],
        "min_samples_split": [2, 5],
    }

grid = GridSearchCV(
    estimator=tuning_model,
    param_grid=param_grid,
    scoring="r2",
    cv=3,
    n_jobs=-1,
    verbose=1,
)
grid.fit(X_train, y_train)

print("\nBest hyperparameters found:")
for k, v in grid.best_params_.items():
    print(f"  {k:<18} = {v}")
print(f"Best CV R-squared (inside grid): {grid.best_score_:.4f}")

tuned_model = grid.best_estimator_
tuned_pred  = tuned_model.predict(X_test)
tuned_rmse  = np.sqrt(mean_squared_error(y_test, tuned_pred))
tuned_mae   = mean_absolute_error(y_test, tuned_pred)
tuned_r2    = r2_score(y_test, tuned_pred)

tuned_cv = cross_val_score(tuned_model, X_train, y_train, cv=kfold, scoring='r2', n_jobs=-1)

print("\n" + "="*60)
print(f"TUNED {best_ensemble_name.upper()} - HOLD-OUT PERFORMANCE")
print("="*60)
print(f"R2 Score     : {tuned_r2:.4f}")
print(f"RMSE         : {tuned_rmse:.4f}")
print(f"MAE          : {tuned_mae:.4f}")
print(f"CV Mean (R2) : {tuned_cv.mean():.4f}")
print(f"CV Std  (R2) : {tuned_cv.std():.4f}")
```

# 8. Combined Comparison - Baselines vs Ensembles

This section combines the baseline results, the ensemble results, and the tuned ensemble into a single table for the direct baseline-vs-ensemble comparison.


```python
tuned_row = pd.DataFrame([{
    "Model":    f"{best_ensemble_name} (tuned)",
    "R2 Score": round(tuned_r2, 4),
    "RMSE":     round(tuned_rmse, 4),
    "MAE":      round(tuned_mae, 4),
    "CV Mean (R2)": round(tuned_cv.mean(), 4),
    "CV Std (R2)":  round(tuned_cv.std(), 4),
}])

all_results = pd.concat([df_base, df_ens, tuned_row], ignore_index=True)
all_results = all_results.sort_values(by="R2 Score", ascending=False).reset_index(drop=True)

def tag(m):
    if m in df_base["Model"].values: return "Baseline"
    if m == f"{best_ensemble_name} (tuned)": return "Ensemble (tuned)"
    return "Ensemble"
all_results.insert(1, "Category", all_results["Model"].apply(tag))

print("="*70)
print("BASELINE vs ENSEMBLE - FINAL COMPARISON")
print("="*70)
print(all_results.to_string(index=False))

metrics_plot = all_results.set_index("Model")[["R2 Score", "RMSE", "MAE", "CV Mean (R2)"]]
x = np.arange(len(metrics_plot.index))
width = 0.2

fig, ax = plt.subplots(figsize=(14, 6))
for i, col in enumerate(metrics_plot.columns):
    ax.bar(x + i * width, metrics_plot[col], width, label=col)
ax.set_xticks(x + width * 1.5)
ax.set_xticklabels(metrics_plot.index, rotation=25, ha="right")
ax.set_ylabel("Metric value")
ax.set_title("Baseline vs Ensemble Regression Models - Metric Comparison",
             fontsize=14, fontweight="bold")
ax.legend(loc="best")
ax.grid(True, axis="y", linestyle="--", alpha=0.4)
plt.tight_layout()
plt.show()

overall_best_name = all_results.iloc[0]["Model"]
print("\n" + "-"*60)
print(f"OVERALL BEST MODEL: {overall_best_name}")
print("-"*60)

if overall_best_name == f"{best_ensemble_name} (tuned)":
    final_model = tuned_model
    final_pred  = tuned_pred
elif overall_best_name in ensemble_grp:
    final_model = ensemble_grp[overall_best_name]
    final_pred  = final_model.predict(X_test)
else:
    final_model = baseline_grp[overall_best_name]
    if overall_best_name in ["Linear Regression", "KNN"]:
        final_pred = final_model.predict(X_test_scaled)
    else:
        final_pred = final_model.predict(X_test)
```

- **Modest ensemble gap.** The ~3 R-squared point improvement from tuned XGBoost over a single Decision Tree is real but not dramatic. For use cases where interpretability is highly prized, the Decision Tree baseline may be a defensible choice.
- **Label encoding of `country`** imposes an artificial ordering that does not exist in reality. Tree-based models handle this correctly, but for richer linear baselines target encoding or frequency encoding could be tried.
- **No time-series treatment.** The `year` column is treated as an ordinary feature rather than as a temporal index. A time-based train/test split and a rolling retraining schedule would be appropriate for production use, where concept drift in work patterns is a real risk.
- **Local interpretability.** Global feature importance gives population-level insight but does not explain individual predictions. SHAP values would close this gap.
- **Hyperparameter grid could be wider.** The GridSearchCV grid was deliberately small to keep runtime manageable. Bayesian optimisation over a wider grid is likely to yield small additional gains.

Overall, the project has demonstrated that boosted-tree ensembles - and tuned XGBoost in particular - reliably outperform linear and distance-based baselines on this type of workforce data, and that **wellbeing features, not workload features, are the dominant predictors of employee productivity** in remote and hybrid work environments.
